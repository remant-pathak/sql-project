```python
import sqlite3
```

```python
conn= sqlite3.connect("database.db")
cursor=conn.cursor()
```

```python
cursor.execute(''' 
    insert into student (name,phone,address,grade)
    values
    ('ram','123456789','kathmandu','3'),
    ('shyam','9875568445','lalitpur','6'),
    ('hari','9898888','janakpur','66');

''')

conn.commit()
print("student data inserted succesfully")
```

```python
cursor.execute(''' 
  select * from student;
''')
data = cursor.fetchall()
for row in data:
    print(row)
    
```

```python
cursor.execute(''' 
update student set phone='9889898985858' where student_id=1;
''')
conn.commit() # to change data permanently
print("student changed succesfully")
```

```python
cursor.execute('''
delete from student where student_id=1;
''')
conn.commit()
print("student deleted succesfully.")
```
